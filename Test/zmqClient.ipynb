```python
import zmq
import time
import json
```

```python
print(zmq.zmq_version())    # 查看ZeroMQ库版本
print(zmq.pyzmq_version())  # 查看pyzmq版本
```

```python
# 客户端
context = zmq.Context()
client_socket = context.socket(zmq.SUB)
client_socket.connect("tcp://localhost:5555")
    
```

```python
# 订阅者
client_socket.setsockopt_string(zmq.SUBSCRIBE,"news")
for i in range(10):
    message = client_socket.recv_string()
    print(f"Received: {message}")
```

```python
# 发送数据
for req in range(10):
    #socket.send_string(f"hello {req}")
    client_socket.send_json()
    message = client_socket.recv_json()
    print(f"recvived reply:{message}")
```
