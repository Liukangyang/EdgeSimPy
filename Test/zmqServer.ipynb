```python
import zmq
import time
```

```python
print(zmq.zmq_version())    # 查看ZeroMQ库版本
print(zmq.pyzmq_version())  # 查看pyzmq版本
```

```python
#服务端
context = zmq.Context()
server_socket = context.socket(zmq.PUB)
server_socket.bind("tcp://*:5555")
    
```

```python
# 发布方
topic = "news"
for i in range(10):
    message = f"{topic} Message {i}"
    server_socket.send_string(message)
    time.sleep(1)
    
```

```python

# 服务端接收数据
while True:
    message = server_socket.recv_string()
    print(f"recv message:{message}")
    server_socket.send_string(f"World from {message}")
    
```
